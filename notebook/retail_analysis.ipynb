```python
# This Python 3 environment comes with many helpful analytics libraries installed
# It is defined by the kaggle/python Docker image: https://github.com/kaggle/docker-python
# For example, here's several helpful packages to load

""" 1. Introduction

## Project Overview

This project focuses on analysing retail sales data to uncover insights about:

- Sales performance
- Customer behaviour
- Product performance
- Revenue trends
- Regional performance
- Customer segmentation

The cleaned and enriched dataset will be exported and connected to Power BI to create an interactive dashboard.

## Business Objectives

The analysis aims to answer:

1. How are sales performing over time?
2. Which products and categories generate the highest revenue?
3. Who are the most valuable customers?
4. Which regions contribute the most sales?
5. What customer trends can improve business decisions?

## Dashboard Outcome

The final dataset will support a Power BI dashboard containing:

- Sales KPIs
- Revenue trends
- Customer analytics
- Product insights
- Regional analysis

"""
# 2. Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 

# Input data files are available in the read-only "../input/" directory
# For example, running this (by clicking run or pressing Shift+Enter) will list all files under the input directory

df = pd.read_csv('/kaggle/input/datasets/migueribau/retail-sales/Data (2).csv')

df.head()

df.info()

df.describe()

df.isnull().sum()

#Remove missing customers
df = df.dropna(subset=['Customer ID'])

# Remove cancelled orders
df = df[df['Quantity'] > 0]

# Convert date type
df['InvoiceDate'] = pd.to_datetime(
    df['InvoiceDate'],
    dayfirst=True,
    format='mixed',
    errors='coerce'
)

# Remove invalid dates

df = df.dropna(
    subset=['InvoiceDate']
)

df['InvoiceDate'].isna().sum()

# Create revenue column
df['Revenue'] = df['Quantity'] * df['Price']

# Create month column
df['Month'] = df['InvoiceDate'].dt.month

# Create year column
df['Year'] = df['InvoiceDate'].dt.year

### Analysis

# Total revenue
total_sales = df['Revenue'].sum()

total_sales

# Monthly sales

monthly_sales = (
    df.groupby('Month')
    ['Revenue']
    .sum()
    .sort_values()
)

monthly_sales

# Top customers
top_customers = (
df.groupby('Customer ID')
['Revenue']
.sum()
.sort_values(ascending=False)
.head(10)
)

top_customers

# Country performance

country_sales = (
df.groupby('Country')
['Revenue']
.sum()
.sort_values(ascending=False)
)

df.columns


# Export for Power BI
df.to_csv(
"clean_retail_data.csv",
index=False
)

# You can write up to 20GB to the current directory (/kaggle/working/) that gets preserved as output when you create a version using "Save & Run All" 
# You can also write temporary files to /kaggle/temp/, but they won't be saved outside of the current session

# Use the kagglehub client library to attach Kaggle resources like competitions, datasets, and models to your session
# Learn more about kagglehub: https://github.com/Kaggle/kagglehub/blob/main/README.md

import kagglehub
# kagglehub.dataset_download('<owner>/<dataset-slug>')
```
